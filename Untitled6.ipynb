```python
loader = PyPDFLoader("My_Resume (Vasanth).pdf")
documents = loader.load()
```

```python
import os
print(os.getcwd())
```

```python

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

docs = text_splitter.split_documents(documents)
```

```python

from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
vectorstore = Chroma.from_documents(docs if len(docs) > 0 else loader.load(), embeddings)
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python

prompt = ChatPromptTemplate.from_template("""
Answer the question based only on the context below.

Context:
{context}

Question:
{question}
""")
```

```python
retriever = vectorstore.as_retriever()
```

```python
rag_chain = (
    {"context": retriever, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
response = rag_chain.invoke(
    "What is the skills mentioned in Vasanth_Resume?"
)
print(response)
```

```python

response = rag_chain.invoke(
    "What is the Number mentioned in that resume?"
)
print(response)
```

```python
response = rag_chain.invoke(
    "What is the Education and Projects mentioned in that resume?"
)
print(response)
```
